Pandas-DataFrame and Series

Pandas is a powerful data manipulation library in python, widely used for "Data analysis and Data cleaning". It provides two primary data structure: Series and DataFrame. A Series is a 1D array like object, while a DataFrame is a 2D, size-mutable and potentially heterogeneous tabular data structure with labled axes(row and columns).

```python
import pandas as pd
```

```python
# Series
# Pandas Series is a 1-D array-like object that can hold any data type. It is similar to a column in a spreadsheet or a SQL table.
data = [1, 2, 3, 4, 5]
series = pd.Series(data)
print("Series \n", series)
print(type(series))

```

```python
# create a Series from a dictionary
data = {'a': 10, 'b': 20, 'c': 30} # a, b, c are indices and 10, 20, 30 are values
series_dict = pd.Series(data) 
print("Series from dictionary \n", series_dict) 
```

```python
data = [10, 20, 30, 40, 50]
indexes = ['a', 'b', 'c', 'd', 'e']
pd.Series(data, index=indexes)
```

DataFrame

```python
# DataFrame
# Create a DataFrame from a dictionary of lists
data = { 'Name': ['Alice', 'Bob', 'Charlie', 'David'],
         'Age': [24, 27, 22, 32],
         'City': ['New York', 'Los Angeles', 'Chicago', 'Houston'] 
       }
df = pd.DataFrame(data)
print("DataFrame \n", df)
print(type(df))
```

```python
# Create a DataFrame from a list of dictionaries
data = [ {'Name': 'Alice', 'Age': 24, 'City': 'New York'},
         {'Name': 'Bob', 'Age': 27, 'City': 'Los Angeles'},
         {'Name': 'Charlie', 'Age': 22, 'City': 'Chicago'},
         {'Name': 'David', 'Age': 32, 'City': 'Houston'} 
       ]
df=pd.DataFrame(data)
print("DataFrame from list of dictionaries \n", df)
print(type(df))
```

Data Manipulation and Analysis with Pandas

Data Manipulation and Analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of function for data manipulation and analysis, making it easier to clean, transform and extract insights from data.

```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
```

```python
df.head(5)  # Display first 5 rows of the DataFrame
```

```python
# Handling Missing Data
df.isnull()  # Check for missing values in the DataFrame
```

```python
df.isnull().any()  # Check if any column has missing values
```

```python
df.isnull().sum()  # Get count of missing values in each column
```

```python
df.fillna(0)  # Fill missing values with 0
```

```python
df_filled = df.fillna(0)
```

```python
# Filling missing values with the mean of the column
df['Sales_FillNA'] = df['Sales'].fillna(df['Sales'].mean()) # Fill missing values in 'Sales' column with the mean of that column
df
```

```python
# Rename columns
df = df.rename(columns = {'Date': 'Sales Date'}) # Rename 'Date' column to 'Sales Date'
df.head(5)
```

```python
# Change data type of a column
df['Value_new'] = df['Value'].fillna(df['Value'].mean()).astype(int) # Fill missing values in 'Value' column with the mean and convert to integer
df.head(5)
```

```python
df['New Value'] = df['Value'].apply(lambda x:x*2) # Multiply each value in 'Value' column by 2 and create a new column 'New Value'
df.head(5)
```

```python
# Data Aggregation and Grouping
grouped_mean = df.groupby('Product')['Value'].mean()  # Calculate mean of 'Value' column grouped by 'Product'
grouped_mean
```

```python
grouped_sum = df.groupby(['Product', 'Region'])['Value'].sum()  # Calculate sum of 'Value' column grouped by 'Product' and 'Region'
grouped_sum
```

```python
# Aggregation with multiple functions
grouped_agg = df.groupby(['Region'])['Value'].agg(['mean', 'sum', 'count'])  # Calculate mean, sum, count of 'Value' column grouped by 'Region'
grouped_agg
```
